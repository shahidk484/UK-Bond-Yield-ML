```python
file_path = '../data/raw/GLC Nominal daily data current month.xlsx'

df = pd.read_excel(
    file_path,
    sheet_name='4. spot curve' 
)
print(df.shape)
display(df.head())
```

```python
print(df.iloc[2 :])
```

```python
for column_number, maturity in enumerate(df.iloc[2, 1:]):
    print(column_number + 1, maturity)
```

```python
dates = df.iloc[4:, 0]
yield_10y = df.iloc[4:, 20]
print(yield_10y) 
```

```python
yield_data = pd.DataFrame({
    'Date': dates,
    'Yield_10Y': yield_10y
})
yield_data = yield_data.reset_index(drop=True)

display(yield_data)
```

```python
print('Number of observations:', len(yield_data))

print('Missing values:')
print(yield_data.isnull().sum())

print('data types:')
print(yield_data.dtypes)
```
